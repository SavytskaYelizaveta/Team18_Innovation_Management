# THE EPO TECHNOLOGY INTELLIGENCE PLATFORM (TIP)

# Introduction

The Technology Intelligence Platform is a cloud-based software development environment built upon the **Jupyter ecosystem** and powered by the **Python** programming language. Specifically, it leverages **JupyterLab** for its rich web interface and JupyterHub to manage its multi-user capabilities.

The platform has a data-centric design: it provides users with direct access to patent and prior art databases (based either on SQL access to BigQuery datasets or on calls to RESTful APIs) wherein data retrieval is streamlined by a set of pre-installed Python libraries developed by the EPO. This integrated approach simplifies the process of retrieving, filtering and processing data. Users can then use the platform to develop customized solutions for statistics or general data analysis, create customized dashboards, forecasting models, trend analysis and more.



## Reminder

1) What is **jupypter lab**? JupyterLab is a web-based application providing a rich IDE (Integrated Development Environment) for developing and executing Python code. It is ultimately the interface of the Technology Intelligence Platform to the user.

2) The core concept of the JupyterLab application is the Jupyter Notebook, which is a document containing live code, equations, visualizations, and narrative text. The Jupyter Notebook hosts and organizes such content in a sequence of cells.

## Server shutdown/restart: File > Hub control panel
The user server can be stopped if not needed anymore to release resources to other users, and it can be subsequently restarted to resume activity. A stop/restart can be very helpful as first intervention in case of troubles, e.g. not anymore responsive, overloading or unclear misbehavior.



## Extensions

Round button in the left vertical menu. For instance: Git extension, more explanation available at https://github.com/jupyterlab/jupyterlab-git

## Resource usage: CPU, MEM, DISK (top right of the page)

The four monitored resources are:

- The CPU cores usage (4 CORES)

- The amount of allocated memory (RAM) (32 Giga)

- The amount of disk space used in the home folder (30 Giga)

- The amount of downloaded data from the server to the local machine running the browser in the past seven days

## Important notes on reaching resource limit

If the requested CPU resources exceed available capacity, the excess request will be ignored.

If the memory reaches or crosses 100% of available value, the relevant processes will crash or hang

Disk usage reaching 100% can severely impair the functionality of the user server and even prevent it from rebooting!.

In case of this event it is recommended to CLEAN UP IMMEDIATELY SOME DISK SPACE and precautionally reboot the server!

Once the download traffic volume limit has been reached, the download speed will be reduced to a drastically low value which will merely guarantee the minimal operativity of the JupyterLab GUI.

## Terminal
![tip](img/terminal.png)

The terminal offers access to a wide range of basic and advanced Unix commands, enabling efficient system management and functionality access that may not be available through the GUI. This includes in particular file manipulation and process management. But even more important, from the terminal users can manage and configure their working environment using preinstalled package management tools such as Pip and Conda. These tools allow for the creation and management of the powerful “virtual environments”, as well as the installation, updating, and removal of software packages, ensuring that the development environment is tailored to specific activitiy needs. The terminal also enables using all the broad features of the Git versioning tool, and it facilitates authentication and interaction with remote repository management systems like GitHub, enabling seamless version control and collaboration.
